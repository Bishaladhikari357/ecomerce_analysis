```python
categorical_features = [
    # City
    'City_Biratnagar',
    'City_Butwal',
    'City_Chitwan',
    'City_Kathmandu',
    'City_Lalitpur',
    'City_Pokhara',

    # Product
    'Product_Headphones',
    'Product_Keyboard',
    'Product_Laptop',
    'Product_Monitor',
    'Product_Mouse',
    'Product_Printer',
    'Product_Smartphone',
    'Product_Tablet',

    # Category
    'Category_Accessories',
    'Category_Audio',
    'Category_Electronics',
    'Category_Office',

    # Brand
    'Brand_Apple',
    'Brand_Dell',
    'Brand_HP',
    'Brand_LG',
    'Brand_Logitech',
    'Brand_Samsung',
    'Brand_Sony',

    # Payment Method
    'Payment_Method_Bank Transfer',
    'Payment_Method_Card',
    'Payment_Method_Cash',
    'Payment_Method_Khalti',
    'Payment_Method_eSewa',

    # Shipping Method
    'Shipping_Method_Express',
    'Shipping_Method_Standard',

    # Order Status
    'Order_Status_Cancelled',
    'Order_Status_Delivered',
    'Order_Status_Processing',
    'Order_Status_Shipped'
]
```

```python
from scipy.stats import chi2_contingency 
import pandas as pd 
alpha =0.05
df_cleaned['Total_Sales_bin'] = pd.qcut(df_cleaned['Total_Sales'],q = 4, labels = False)
chi2_results = {}

for col in categorical_features:
    contingency = pd.crosstab(df_cleaned[col],df_cleaned['Total_Sales_bin'])
    chi2_stat, p_val, _, _ =chi2_contingency(contingency)
    decision = 'reject Null (keep Feature)' if p_val < alpha else  'Accept Null (Drop Feature)'
    chi2_results[col] = {
    'chi2_statistic': chi2_stat,
    'p_value': p_val,
    'Decision': decision
    }

chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by = 'p_value')
chi2_df

    
```
